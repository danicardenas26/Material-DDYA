## 1) Descripcion del problema **[1 pts]**

El parcial consiste en desarrollar un sistema para una empresa de transporte urbano que permita representar, gestionar y recorrer las conexiones directas entre varias estaciones de su red. El sistema se fundamenta en grafos, modelando la red como un grafo no dirigido donde cada estacion es un vertice y cada conexion es una arista, almacenadas de forma bidireccional mediante una lista de adyacencia.

### Se nos pide:

1. **Representacion de la red de estacion:** Construir el grafo utilizando una lista de adyacencia e imprimir cada estacion con sus conexiones directas.

2. **Exploracion BFS (Anchura):** Implementar el recorrido BFS iniciando por la estacion "Central", utilizando una cola, y mostrando el orden de visita junto con el grado de separacion de cada nodo respecto al origen.

3. **Exploracion DFS (Profundidad):** Implementar el recorrido DFS iniciando por la estacion "Central", avanzando por una rama hasta donde sea posible antes de regresar.

4. **Seleccion dinamica de inicio:** Permitir que el usuario indique desde que estacion desea iniciar los recorridos BFS y DFS (con validacion de errores si la estacion no existe).

5. **Ruta mas corta entre dos estaciones:** Solicitar una estacion inicial y una destino, y utilizando BFS, encontrar la ruta con la menor cantidad de conexiones (saltos) entre ambas.

6. **Actualizacion de la red:** Permitir el registro de una nueva estacion en la red y conectarla con al menos dos estaciones previamente existentes.

### Condiciones Tecnicas:

* El grafo debe representarse mediante una **lista de adyacencia**.

* El grafo es **no dirigido**, por lo que cada conexion debe registrarse en ambos sentidos sin duplicar datos.

* El BFS debe implementarse obligatoriamente con una **cola**.

* El DFS debe implementarse mediante **recursividad o utilizando una pila**.

* Cada vertice debe visitarse una sola vez por recorrido, apoyandose en una estructura de **vertices visitados** (ej. un `set`).

* **No esta permitido** el uso de librerias externas que implementen directamente los algoritmos BFS o DFS.

### 5) Analisis de complejidad **[15 pts]**
Definiendo V como el numero de vertices (estaciones) y E como el numero de aristas (conexiones):
#### Justificacion:
Al utilizar una lista de adyacencia para representar el grafo, tanto el algoritmo BFS como el DFS visitan cada vertice exactamente una vez y evaluan cada arista conectada.

- Mejor Caso: O(1) -> Esto ocurre en la funcion de "Encontrar Ruta" si la estacion de origen y destino son la misma, o si son adyacentes y es la primera arista evaluada en la lista.




- Peor Caso: O(V + E) **"(O(n))"** -> Ocurre en la generacion de los recorridos completos de BFS o DFS, o si el destino en la busqueda de la ruta se encuentra en el ultimo nivel del grafo.




#### Funcion de complejidad T(n):
T(V, E) = O(V + E). Si simplificamos asumiendo que el tamaño general de la entrada es n = V + E, la funcion final es:

- T(n) = O(n).

## 6) Código funcional **[15 pts]**

```python
# 1. Representacion de la red de estaciones: Grafo inicial usando lista de adyacencia 
grafo_estaciones = {
   "Central": ["Norte", "Museo"],
   "Norte": ["Central", "Parque", "Universidad"],
   "Museo": ["Central", "Hospital"],
   "Parque": ["Norte"],
   "Universidad": ["Norte", "Terminal"],
   "Hospital": ["Museo", "Terminal"],
   "Terminal": ["Universidad", "Hospital"]
}

def imprimir_grafo(grafo):
   print("\n--- Red de Estaciones Actual ---")
   for estacion, conexiones in grafo.items():
       print(f"{estacion}: {', '.join(conexiones)}")
   print("--------------------------------\n")

# 2. Exploracion BFS

def recorrido_bfs(grafo, inicio):
    # Validamos si el usuario existe
    if inicio not in grafo:
        print(f"Error: La estacion '{inicio}' no existe en la red.")
        return
    
    visitados = set()
    cola = [(inicio, 0)] # La cola almacena tuplas: (nodo_actual, grado_separacion)
    visitados.add(inicio)
    
    recorrido_orden = []
    grados_dict = {} # Para agrupar a las personas por grado
    
    while cola:
        # pop(0) simula el comportamiento de una cola en una lista FIFO
        nodo_actual, grado = cola.pop(0)
        recorrido_orden.append(nodo_actual)
        
        # Agrupa por grado
        if grado not in grados_dict:
            grados_dict[grado] = []
        grados_dict[grado].append(nodo_actual)
        
        # Explora vecinos
        for vecino in grafo[nodo_actual]:
            if vecino not in visitados:
                visitados.add(vecino)
                cola.append((vecino, grado + 1))
                
    print(f"\nRecorrido BFS desde {inicio}:")
    print(" -> ".join(recorrido_orden))
    for g, personas in grados_dict.items():
        print(f"Grado {g}: {', '.join(personas)}")

# 3. Exploracion DFS
def recorrido_dfs(grafo, nodo_actual, visitados=None, recorrido_orden=None):
    # Inicializacion en la primera llamada recursiva
    if visitados is None:
        visitados = set()
        recorrido_orden = []
        
        # Validacion de usuario inicial
        if nodo_actual not in grafo:
            print(f"Error: La estacion '{nodo_actual}' no existe en la red.")
            return []

    # Marca el nodo como visitado y lo agregamos al recorrido
    if nodo_actual not in visitados:
        visitados.add(nodo_actual)
        recorrido_orden.append(nodo_actual)
        
        # Llamada recursiva para avanzar por la rama
        for vecino in grafo[nodo_actual]:
            if vecino not in visitados:
                recorrido_dfs(grafo, vecino, visitados, recorrido_orden)
                
    return recorrido_orden

# 4. Se desarrolla en MAIN
# 5. Ruta mas corta entre dos estaciones:

def ruta_mas_corta(grafo, inicio, destino):
    if inicio not in grafo or destino not in grafo:
        print("Error: La estacion inicial o la destino no existen.")
        return
    
    # La cola almacena RUTAS completas (listas), no solo nodos
    cola = [[inicio]]
    visitados = set()
    visitados.add(inicio)
    
    while cola:
        ruta_actual = cola.pop(0)
        nodo_actual = ruta_actual[-1] # El ultimo nodo de la ruta evaluada
        
        # Si llegamos al destino, terminamos
        if nodo_actual == destino:
            grados_separacion = len(ruta_actual) - 1
            print(f"\nRuta encontrada desde {inicio} hasta {destino}:")
            print(" -> ".join(ruta_actual))
            print(f"Grados de separacion: {grados_separacion}")
            return ruta_actual
        
        # Exploramos vecinos
        for vecino in grafo[nodo_actual]:
            if vecino not in visitados:
                visitados.add(vecino)
                # Copia de la ruta actual y agregamos el vecino
                nueva_ruta = list(ruta_actual)
                nueva_ruta.append(vecino)
                cola.append(nueva_ruta)
                
    # Si la cola se vacia y no retorno, no hay conexion
    print(f"\nNo existe una conexion entre la estacion {inicio} y la estacion {destino}.")
    return None

# 6. Agregar una nueva estacion y sus debidas conexiones:

def agregar_estacion(grafo, nueva_estacion, conexiones):
    # Si no existe creamos su lista vacia en el diccionario
    if nueva_estacion not in grafo:
        grafo[nueva_estacion] = []
        
    for conexion in conexiones:
        if conexion in grafo:
            # Agrega la conexion en ambos sentidos para que sea no dirigido
            if conexion not in grafo[nueva_estacion]:
                grafo[nueva_estacion].append(conexion)
            if nueva_estacion not in grafo[conexion]:
                grafo[conexion].append(nueva_estacion)
    print(f"\n Estacion '{nueva_estacion}' agregada y conectada con {', '.join(conexiones)}")

# MAIN:
def main():
    while True:
        print("\n" + "=====================")
        print("MENU DE RED DE TRANSPORTE")
        print("=====================")
        print("1. Mostrar red de estaciones")
        print("2. Ejecutar recorrido BFS")
        print("3. Ejecutar recorrido DFS")
        print("4. Buscar ruta mas corta entre estaciones")
        print("5. Agregar nueva estacion y sus conexiones")
        print("6. Salir")
        print("=====================")
        opcion = input("Seleccione una opcion (1-6): ")
        if opcion == "1":
            imprimir_grafo(grafo_estaciones)
        elif opcion == "2":
            estacion = input("Ingrese la estacion inicial para BFS (Ingrese el nombre tal cual ejp. Central): ")
            recorrido_bfs(grafo_estaciones, estacion)
        elif opcion == "3":
            estacion = input("Ingrese la estacion inicial para DFS (Ingrese el nombre tal cual ejp. Central): ")
            resultado = recorrido_dfs(grafo_estaciones, estacion)
            if resultado:
                print(f"\nRecorrido DFS desde {estacion}:\n{' -> '.join(resultado)}")
        elif opcion == "4":
            inicio = input("Ingrese la estacion de origen (Ingrese el nombre tal cual ejp. Central): ")
            destino = input("Ingrese la estacion de destino (Ingrese el nombre tal cual ejp. Parque): ")
            ruta_mas_corta(grafo_estaciones, inicio, destino)
        elif opcion == "5":
            nueva_nombre = input("Ingrese el nombre de la nueva estacion: ")
            conexiones_str = input(f"Ingrese las conexiones de {nueva_nombre} separadas por coma (ej. Terminal, Hospital): ")
            conexiones_sin_espacion = conexiones_str.replace(" ", "")
            conexiones_lista = conexiones_sin_espacion.split(",")
            agregar_estacion(grafo_estaciones, nueva_nombre, conexiones_lista)
        elif opcion == "6":
            print("\nSaliendo del programa...")
            break
        else:
            print("\nError: Opcion no valida. Por favor, intenta de nuevo.")
main()
```
