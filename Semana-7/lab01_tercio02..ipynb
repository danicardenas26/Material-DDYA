## 1) Descripción del problema **[1 pts]**

Una tienda minorista enfrenta problemas en la gestión de sus procesos diarios, lo que genera retrasos y una mala experiencia tanto para los empleados como para los clientes. Actualmente, no existe un control claro del orden de llegada de los clientes, no hay un historial que permita corregir errores humanos (como saltarse un turno o registrar una atención indebida), y el catálogo de productos se maneja de forma desorganizada, dificultando la búsqueda y actualización de precios.

Para resolver esto, se requiere un sistema unificado que estructure el flujo de trabajo mediante tres pilares computacionales:

1. Gestión de turnos (Cola): Para garantizar que la atención siga un orden justo (FIFO: primero en entrar, primero en salir).

2. Historial de operaciones (Pila): Para proveer un mecanismo de "deshacer" en caso de error (LIFO: último en entrar, primero en salir).

3. Catálogo dinámico (Lista Doblemente Enlazada): Para permitir una navegación bidireccional ágil del inventario y facilitar la inserción y eliminación de productos en tiempo real.

## 2) Requerimientos **[2 pts]**

### Requerimientos Funcionales (RF)

- RF1 (Gestión de Cola): El sistema debe permitir inicializar una cola con n clientes.

- RF2 (Gestión de Cola): El sistema debe permitir agregar nuevos clientes al final de la cola y atender (remover) al cliente que esté al frente.

- RF3 (Gestión de Cola): El sistema debe mostrar el estado actual de los clientes que esperan.

- RF4 (Gestión de Pila): Cada vez que se atiende a un cliente, el sistema debe registrar automáticamente su nombre en el tope de una pila.

- RF5 (Gestión de Pila): El sistema debe permitir "deshacer" la última atención. Esto retirará al cliente del tope de la pila y lo reinsertará al inicio de la cola.

- RF6 (Gestión de Catálogo): El sistema debe permitir agregar productos (código, nombre, precio) al catálogo.

- RF7 (Gestión de Catálogo): El sistema debe permitir listar los productos del primero al último y del último al primero.

- RF8 (Gestión de Catálogo): El sistema debe permitir buscar y eliminar un producto específico ingresando su código.

### Requerimientos No Funcionales (RNF)

- RNF1 (Estructuras de Datos): Es obligatorio implementar las estructuras de Cola, Pila y Lista Doblemente Enlazada desde cero (sin usar librerías de alto nivel que ya las resuelvan).

- RNF2 (Interfaz): El sistema debe interactuar con el usuario mediante un menú de consola (CLI).

- RNF3 (Complejidad): Las operaciones de inserción en la pila y la cola deben ejecutarse en tiempo constante.

## 4) Diagramas de flujo **[5 pts]**




El flujo arranca en el nodo Inicio y desemboca en un Menú Principal con forma de rombo (que indica una decisión o selección), desde aquí, el usuario tiene tres caminos:

- Opción 1: Ingresa al módulo de clientes.

- Opción 2: Ingresa al módulo de catálogo.

- Opción 3 (Salir): Termina la ejecución del sistema.

Módulo Clientes (Gestión de Colas y Pilas): Este submódulo maneja la recepción y atención de clientes combinando dos estructuras para simular una sala de espera con opción de deshacer errores.

1. Agregar cliente: Inserta a la persona en una Cola (el primero en llegar es el primero en ser atendido).

2. Atender cliente: Retira al cliente que está al frente de la Cola. Automáticamente activa la sub-rutina Guardar en Pila, creando un historial temporal de los clientes que ya fueron procesados.

3. Ver clientes en espera: Muestra a los usuarios que siguen en la Cola.

4. Deshacer última atención: Permite corregir un error si se atendió a alguien por equivocación. Ejecuta Extraer de Pila (saca al último cliente guardado en el historial) y luego hace un Insertar Inicio Cola para devolver a esa persona al frente de la fila de espera.

Módulo Catálogo (Lista Doblemente Enlazada): Este submódulo está diseñado para gestionar un inventario o catálogo. Utiliza una Lista Doble, lo cual significa que cada elemento (nodo) tiene una conexión tanto con el elemento siguiente como con el anterior, permitiendo la navegación bidireccional.

1. Agregar producto: Inserta un nuevo nodo/producto en la lista.

2. Mostrar adelante: Recorre y muestra los productos desde el inicio hasta el final de la lista.

3. Mostrar atrás: Recorre y muestra los productos en sentido inverso, desde el final hacia el inicio.

4. Buscar producto: Localiza un elemento específico dentro del catálogo.

5. Eliminar producto: Borra un elemento y reajusta automáticamente los enlaces dobles de la lista para no perder la continuidad.

## 5) Diagramas de secuencia **[5 pts]**




Acción 1: El usuario selecciona la opción 2 ("Atender Cliente").

Extracción (Cola): El sistema va a la Cola y saca al cliente que está en la posición 0 (el primero de la fila).

Guardado (Pila): El sistema toma a ese cliente y lo inserta al final de la Pila (el tope del historial).

Acción 2: El usuario selecciona la opción 4 ("Deshacer última atención").

Recuperación (Pila): El sistema va a la Pila y saca al último cliente que fue guardado en el tope.

Restauración (Cola): El sistema toma a ese cliente recuperado y lo inserta a la fuerza en la posición 0 de la Cola. El cliente vuelve a ser el primero en esperar.

## 7) Análisis de complejidad **[5 pts]**

1. Inserción de un Producto en el Catálogo

- Operación: Agregar un producto al final de la lista doblemente enlazada.

- Función T(n): T(n) = c (una constante, donde c representa el tiempo de asignación de punteros).

- Complejidad: O(1) (Tiempo constante).

- Justificación: Gracias a que mantenemos una referencia directa a la cola (catalogo['cola']), no necesitamos recorrer toda la lista para insertar un nuevo elemento. Los apuntadores siguiente y anterior se actualizan de manera inmediata sin importar cuántos productos existan previamente.

2. Búsqueda y Eliminación en el Catálogo

- Operación: Buscar un producto por su código o eliminarlo.

- Función T(n): $T(n) = a \cdot n + b$ (donde n es el número total de productos en el catálogo).

- Complejidad: O(n) (Tiempo lineal).

- Justificación: Al tratarse de una estructura enlazada sin índices numéricos directos, el algoritmo se ve obligado a recorrer los nodos uno por uno (utilizando el campo siguiente) en el peor de los casos hasta el final de la lista para encontrar o descartar el código buscado.

3. Operaciones en la Cola de Clientes (Atender / Registrar)

- Operación: Atender al primer cliente de la cola (cola_clientes = cola_clientes[1:]).

- Función T(n): $T(n) = a \cdot n + b$ (donde n es la cantidad de clientes en espera).

- Complejidad: O(n) para la extracción y O(1) para la inserción (.append()).

- Justificación: Aunque agregar un elemento al final con .append() toma tiempo constante O(1), la extracción del primer elemento mediante slicing ([1:]) obliga a Python a crear una nueva lista en memoria copiando los n-1 elementos restantes, lo cual escala de forma lineal respecto al tamaño de la cola.

### Graficas:

Mejor caso: O(1) -> Insercion en catalogo




Peor caso: O(n) -> Búsqueda, listado y manejo de la cola





## 8) Código funcional **[10 pts]**


```python
def pedir_entero_positivo(mensaje):
    """Verifica manualmente que el string contenga solo números del 0 al 9."""
    valido = False
    numero = 0
    while not valido:
        entrada = input(mensaje)
        if entrada == "":
            print("Error: el dato no puede estar vacío.")
            continue
        
        es_num = True
        for caracter in entrada:
            if caracter not in "0123456789":
                es_num = False
                break
                
        if es_num:
            numero = int(entrada)
            valido = True
        else:
            print("Error: Ingrese un número entero positivo (sin letras, negativos ni símbolos).")
            
    return numero

def pedir_texto(mensaje):
    """Garantiza que el usuario no ingrese cadenas vacías."""
    valido = False
    texto = ""
    while not valido:
        texto = input(mensaje)
        if texto == "":
            print("Error: el texto no puede estar vacío.")
        else:
            valido = True
    return texto

def agregar_producto(catalogo, codigo, nombre, precio):
    nuevo_nodo = {
        'codigo': codigo, 
        'nombre': nombre, 
        'precio': precio, 
        'siguiente': None, 
        'anterior': None
    }
    
    if catalogo['cabeza'] is None:
        catalogo['cabeza'] = nuevo_nodo
        catalogo['cola'] = nuevo_nodo
    else:
        catalogo['cola']['siguiente'] = nuevo_nodo
        nuevo_nodo['anterior'] = catalogo['cola']
        catalogo['cola'] = nuevo_nodo

def buscar_producto(catalogo, codigo):
    actual = catalogo['cabeza']
    while actual is not None:
        if actual['codigo'] == codigo:
            return actual
        actual = actual['siguiente']
    return None

def eliminar_producto(catalogo, codigo):
    actual = catalogo['cabeza']
    while actual is not None:
        if actual['codigo'] == codigo:
            ant = actual['anterior']
            sig = actual['siguiente']

            if ant is not None:
                ant['siguiente'] = sig
            else:
                catalogo['cabeza'] = sig

            if sig is not None:
                sig['anterior'] = ant
            else:
                catalogo['cola'] = ant
            return True
        actual = actual['siguiente']
    return False

def listar_productos(catalogo, reverso=False):
    if reverso:
        actual = catalogo['cola']
        print("\n--- Catálogo (Z-A) ---")
        while actual is not None:
            print(f"[{actual['codigo']}] {actual['nombre']} - ${actual['precio']}")
            actual = actual['anterior']
    else:
        actual = catalogo['cabeza']
        print("\n--- Catálogo (A-Z) ---")
        while actual is not None:
            print(f"[{actual['codigo']}] {actual['nombre']} - ${actual['precio']}")
            actual = actual['siguiente']

def main():
    cola_clientes = []
    pila_historial = []
    catalogo = {'cabeza': None, 'cola': None}
    
    print("--- INICIO DE COLA ---")
    n_clientes = pedir_entero_positivo("¿Cuántos clientes hay inicialmente en la cola?: ")
    for rep in range(n_clientes):
        nombre = pedir_texto(f"Nombre del cliente {rep+1}: ")
        cola_clientes.append(nombre)
        
    salir = False
    while not salir:
        print("\n=== MENÚ DE LA TIENDA ===")
        print("1. Atender al siguiente cliente")
        print("2. Registrar nuevo cliente (al final de la cola)")
        print("3. Mostrar estado de la cola")
        print("4. Deshacer última atención")
        print("5. Agregar producto al catálogo")
        print("6. Buscar producto")
        print("7. Eliminar producto")
        print("8. Listar catálogo")
        print("9. Salir")
        
        opcion = pedir_entero_positivo("Seleccione una opción: ")
        
        if opcion == 1:
            if len(cola_clientes) > 0:
                cliente_atendido = cola_clientes[0]
                cola_clientes = cola_clientes[1:] # Extraer el primero (FIFO)
                pila_historial.append(cliente_atendido)
                print(f"Atendiendo a: {cliente_atendido}")
            else:
                print("No hay clientes en la cola.")
                
        elif opcion == 2:
            nuevo_cliente = pedir_texto("Nombre del nuevo cliente: ")
            cola_clientes.append(nuevo_cliente)
            print("Cliente agregado a la cola.")
            
        elif opcion == 3:
            print("\nClientes en espera:")
            if len(cola_clientes) == 0:
                print("Cola vacía.")
            else:
                for rep in range(len(cola_clientes)):
                    print(f"{rep+1}. {cola_clientes[rep]}")
                    
        elif opcion == 4:
            if len(pila_historial) > 0:
                ultimo_atendido = pila_historial[-1]
                pila_historial = pila_historial[:-1] # Extraer el último (LIFO)
                cola_clientes = [ultimo_atendido] + cola_clientes # Regresar al inicio
                print(f"Atención deshecha. {ultimo_atendido} vuelve a esperar.")
            else:
                print("No hay atenciones en el historial.")
                
        elif opcion == 5:
            cod = pedir_entero_positivo("Código del producto: ")
            if buscar_producto(catalogo, cod) is not None:
                print("Error: Ya existe un producto con ese código.")
            else:
                nom = pedir_texto("Nombre del producto: ")
                pre = pedir_entero_positivo("Precio del producto (sin decimales): ")
                agregar_producto(catalogo, cod, nom, pre)
                print("Producto agregado con éxito.")
                
        elif opcion == 6:
            cod = pedir_entero_positivo("Código a buscar: ")
            prod = buscar_producto(catalogo, cod)
            if prod is not None:
                print(f"Encontrado: [{prod['codigo']}] {prod['nombre']} - ${prod['precio']}")
            else:
                print("Producto no encontrado.")
                
        elif opcion == 7:
            cod = pedir_entero_positivo("Código a eliminar: ")
            if eliminar_producto(catalogo, cod):
                print("Producto eliminado correctamente.")
            else:
                print("El producto no existe.")
                
        elif opcion == 8:
            print("  1. Del primero al último")
            print("  2. Del último al primero")
            sub_op = pedir_entero_positivo("Seleccione el orden: ")
            if sub_op == 1:
                listar_productos(catalogo, False)
            elif sub_op == 2:
                listar_productos(catalogo, True)
            else:
                print("Opción inválida.")
                
        elif opcion == 9:
            salir = True
            print("Cerrando el sistema...")
        else:
            print("Opción no válida. Intente de nuevo.")
main()
```

#### Código base para benchmarking

```python
import time
from random import randint
import matplotlib.pyplot as plt

# Calcula el tiempo de ejecucion con base a diferentes tamaños de la entrada
# y nos da una lista de resultados, cada resultado contiene el tamaño de la entrada
# y el tiempo de ejecucion
# TENER EN CUENTA ESTAS ENTRADAS PUEDEN VARIAR SEGÚN EL PROBLEMA

def bench(n: int, step: int = 100) -> list[tuple[int, float]]:
    print("Generando datos ...", end="")
    results: list[tuple[int, float]] = []

    # AJUSTAR LAS ENTRADAS SEGÚN EL PROBLEMA
    for size in range(1, n, step):
        # Generar catálogo sintético de tamaño 'size'
        catalogo_bench = {'cabeza': None, 'cola': None}
        for i in range(size):
            agregar_producto(catalogo_bench, i, f"Prod_{i}", randint(100, 5000))
        
        # Medir peor caso de búsqueda O(N): buscar el último elemento insertado
        codigo_buscado = size - 1
        
        ti = time.perf_counter()
        output = buscar_producto(catalogo_bench, codigo_buscado)  # Ajustar a su función
        tf = time.perf_counter()

        delta = (tf - ti) * 1000  # Convertir a milisegundos
        print(".", end="")
        results.append((size, delta))
    return results


# === CODIGO PARA DIBUJAR ===


def dibujar(x, y, dimensiones=(7, 5)):
    assert len(x) == len(y)
    plt.figure(figsize=dimensiones)
    plt.plot(x, y, marker='o', color='b', label='Tiempo de búsqueda (ms)')
    plt.title("Desempeño Teórico vs Empírico - Búsqueda O(N)")
    plt.xlabel("Tamaño del Catálogo (N)")
    plt.ylabel("Tiempo de Ejecución (ms)")
    plt.grid(True)
    plt.legend()
    plt.show()


# === LLAMADO ===
tiempos = bench(1000, step=100)
x_vals = [t[0] for t in tiempos]
y_vals = [t[1] for t in tiempos]

dibujar(x_vals, y_vals)
```
