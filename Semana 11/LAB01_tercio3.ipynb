### DISEÑO

#### Estrategia de Solucion
Primero se creara una funcion que permita al usuario ingresa un numero entero n correspondiente a la cantidad de 
productos para preguntar despues su precio, tras eso se creara un arreglo de tamaño n en el cual se le preguntara al usuario el precio de cada producto.

Posteriormente se preguntara al usuario si quiere ordenar el arreglo de manera ascendente o descendente, se enviara el arreglo como parametro a una funcion que lo ordene dependiendo de la decision de la decision del usuario.

Se retornara el arreglo ordenado y se imprimira en pantalla.

#### Tipo de Algoritmo

Se utilizara Insertion Sort, un algoritmo de ordenamiento que compara los elementos de un arreglo uno por uno, hasta que quede ordenado, su funcionamiento puede verse como el ordenamiento de una baraja de cartas, asume que el primer elemento está ordenado y, uno por uno, toma los siguientes elementos para compararlos de derecha a izquierda con los que ya están en la parte ordenada, desplazando los valores necesarios para abrir un hueco e "insertar" el elemento actual en su posición correcta.

#### Tipo de implementacion

- Iterativa, la solucion implementara ciclos anidados (un for y un while) para recorrer y comparar los precios.
- Incremental, el algoritmo procesa un precio a la vez, haciéndole un espacio y agregándolo a un subarreglo que va creciendo de forma ordenada hasta completar todos los datos.
- Invariante de ciclo: Al inicio de cada iteración $i$, el subarreglo que va desde la posición $0$ hasta $i-1$ contiene los elementos que fueron evaluados hasta el momento, pero ya se encuentran perfectamente ordenados (ya sea de forma ascendente o descendente).

--- JUSTIFICACION ---
- Requisito: El problema exige explícitamente el uso de Insertion Sort, por lo que esta implementación cumple al 100% con los requerimientos técnicos.
- Eficiencia en el mundo real: Si el administrador de la tienda ingresa los precios en un orden que ya está parcialmente ordenado, el algoritmo se vuelve extremadamente rápido (acercándose a su mejor caso computacional O(n)).

#### CODIGO

```python
def ingresar_datos():
    """
    Función que pide al usuario la cantidad de productos y sus precios.
    Utiliza ciclos while simples para validar las reglas sin usar break.
    """
    n = 0 # Inicializamos n en 0 para entrar a while
    while n < 1:
        n = int(input("Ingrese la cantidad de productos a registrar: "))
        if n < 1:
            print("Error: La cantidad de productos debe ser mayor o igual a 1.")

    precios = []
    
    print("\n--- Ingreso de precios ---")
    for rep in range(n):
        precio = -1.0
        while precio < 0:
            precio = float(input(f"Ingrese el precio del producto {rep + 1}: $"))
            if precio < 0:
                print("Error: El precio no puede ser un valor negativo.")
        precios.append(precio)               
    return precios

def insertion_sort(arreglo, orden):
    """
    Función que recibe el arreglo y la opción de ordenamiento,
    y retorna el arreglo ordenado usando el algoritmo Insertion Sort puro.
    """
    n = len(arreglo)
    for i in range(1, n):
        llave = arreglo[i]
        j = i - 1
        # Orden Ascendente
        if orden == 1:
            while j >= 0 and arreglo[j] > llave:
                arreglo[j + 1] = arreglo[j]
                j = j - 1
        # Orden Descendente
        elif orden == 2:
            while j >= 0 and arreglo[j] < llave:
                arreglo[j + 1] = arreglo[j]
                j = j - 1            
        # Inserción del elemento en su posición correcta
        arreglo[j + 1] = llave
    return arreglo

def main():
    arreglo_precios = ingresar_datos()
    print("\n--- Menú de Ordenamiento ---")
    print("1. Orden ascendente (del más económico al más costoso)")
    print("2. Orden descendente (del más costoso al más económico)")
    opcion = 0
    while opcion != 1 and opcion != 2:
        opcion = int(input("Ingrese el número de la opción deseada (1 o 2): "))
        if opcion != 1 and opcion != 2:
            print("Error: Selección no válida. Ingrese 1 o 2.")

    precios_ordenados = insertion_sort(arreglo_precios, opcion)
    print("\n--- Lista de precios organizada ---")
    for i in range(len(precios_ordenados)):
        print(f"$ {precios_ordenados[i]}")
main()
```
