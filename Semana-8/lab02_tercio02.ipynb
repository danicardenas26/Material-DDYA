## 6) Código funcional **[15 pts]**

```python
import matplotlib.pyplot as plt
import networkx as nx

# --- ESTRUCTURAS Y CREACIÓN DE NODOS ---
def crear_nodo(codigo):
    """Crea una estructura de nodo representada como un diccionario."""
    return {
        'codigo': codigo,
        'izquierdo': None,
        'derecho': None,
        'altura': 1
    }

# --- OPERACIONES EN ÁRBOL BINARIO DE BÚSQUEDA (BST) ---
def insertar_bst(nodo, codigo):
    """Inserta un nuevo código en el BST de forma recursiva."""
    if not nodo:
        return crear_nodo(codigo)
    
    if codigo < nodo['codigo']:
        nodo['izquierdo'] = insertar_bst(nodo['izquierdo'], codigo)
    elif codigo > nodo['codigo']:
        nodo['derecho'] = insertar_bst(nodo['derecho'], codigo)
        
    return nodo

# --- FUNCIONES AUXILIARES Y ROTACIONES EN ÁRBOL AVL ---
def obtener_altura(nodo):
    """Devuelve la altura de un nodo o 0 si es nulo."""
    if not nodo:
        return 0
    return nodo['altura']

def obtener_factor_balance(nodo):
    """Calcula el factor de balance de un nodo (altura izquierda - altura derecha)."""
    if not nodo:
        return 0
    return obtener_altura(nodo['izquierdo']) - obtener_altura(nodo['derecho'])

def rotacion_izquierda(nodo_desbalanceado):
    """Realiza una rotación simple a la izquierda en un subárbol."""
    nueva_raiz = nodo_desbalanceado['derecho']
    subarbol_intermedio = nueva_raiz['izquierdo']

    # Reestructuración de enlaces
    nueva_raiz['izquierdo'] = nodo_desbalanceado
    nodo_desbalanceado['derecho'] = subarbol_intermedio

    # Actualización de alturas
    nodo_desbalanceado['altura'] = 1 + max(
        obtener_altura(nodo_desbalanceado['izquierdo']), 
        obtener_altura(nodo_desbalanceado['derecho'])
    )
    nueva_raiz['altura'] = 1 + max(
        obtener_altura(nueva_raiz['izquierdo']), 
        obtener_altura(nueva_raiz['derecho'])
    )

    return nueva_raiz

def rotacion_derecha(nodo_desbalanceado):
    """Realiza una rotación simple a la derecha en un subárbol."""
    nueva_raiz = nodo_desbalanceado['izquierdo']
    subarbol_intermedio = nueva_raiz['derecho']

    # Reestructuración de enlaces
    nueva_raiz['derecho'] = nodo_desbalanceado
    nodo_desbalanceado['izquierdo'] = subarbol_intermedio

    # Actualización de alturas
    nodo_desbalanceado['altura'] = 1 + max(
        obtener_altura(nodo_desbalanceado['izquierdo']), 
        obtener_altura(nodo_desbalanceado['derecho'])
    )
    nueva_raiz['altura'] = 1 + max(
        obtener_altura(nueva_raiz['izquierdo']), 
        obtener_altura(nueva_raiz['derecho'])
    )

    return nueva_raiz

def insertar_avl(nodo, codigo):
    """Inserta un nuevo código en el Árbol AVL y auto-balancea la estructura."""
    # 1. Inserción estándar de un BST
    if not nodo:
        return crear_nodo(codigo)
    elif codigo < nodo['codigo']:
        nodo['izquierdo'] = insertar_avl(nodo['izquierdo'], codigo)
    elif codigo > nodo['codigo']:
        nodo['derecho'] = insertar_avl(nodo['derecho'], codigo)
    else:
        return nodo  # Evita duplicados

    # 2. Actualización de la altura del nodo actual
    nodo['altura'] = 1 + max(
        obtener_altura(nodo['izquierdo']), 
        obtener_altura(nodo['derecho'])
    )

    # 3. Obtención del factor de balance
    balance = obtener_factor_balance(nodo)

    # 4. Casos de desbalanceo y sus respectivas rotaciones
    # Izquierda-Izquierda (LL)
    if balance > 1 and codigo < nodo['izquierdo']['codigo']:
        return rotacion_derecha(nodo)

    # Derecha-Derecha (RR)
    if balance < -1 and codigo > nodo['derecho']['codigo']:
        return rotacion_izquierda(nodo)

    # Izquierda-Derecha (LR)
    if balance > 1 and codigo > nodo['izquierdo']['codigo']:
        nodo['izquierdo'] = rotacion_izquierda(nodo['izquierdo'])
        return rotacion_derecha(nodo)

    # Derecha-Izquierda (RL)
    if balance < -1 and codigo < nodo['derecho']['codigo']:
        nodo['derecho'] = rotacion_derecha(nodo['derecho'])
        return rotacion_izquierda(nodo)

    return nodo

# --- RECORRIDOS EN ÁRBOLES ---
def recorrido_preorden(nodo, resultado):
    """Guarda el recorrido Preorden (Raíz -> Izquierda -> Derecha)."""
    if nodo:
        resultado.append(nodo['codigo'])
        recorrido_preorden(nodo['izquierdo'], resultado)
        recorrido_preorden(nodo['derecho'], resultado)

def recorrido_inorden(nodo, resultado):
    """Guarda el recorrido Inorden (Izquierda -> Raíz -> Derecha)."""
    if nodo:
        recorrido_inorden(nodo['izquierdo'], resultado)
        resultado.append(nodo['codigo'])
        recorrido_inorden(nodo['derecho'], resultado)

def recorrido_postorden(nodo, resultado):
    """Guarda el recorrido Postorden (Izquierda -> Derecha -> Raíz)."""
    if nodo:
        recorrido_postorden(nodo['izquierdo'], resultado)
        recorrido_postorden(nodo['derecho'], resultado)
        resultado.append(nodo['codigo'])

# --- VISUALIZACIÓN GRÁFICA ---
def graficar_arbol(raiz, titulo="Árbol"):
    """Genera la representación gráfica del árbol usando NetworkX y Matplotlib."""
    if not raiz:
        return
        
    grafo = nx.DiGraph()
    posiciones = {}

    def construir_grafo(nodo, x=0, y=0, nivel=1):
        if nodo is not None:
            grafo.add_node(nodo['codigo'])
            posiciones[nodo['codigo']] = (x, y)
            desplazamiento_x = 1 / (2 ** nivel)
            
            if nodo['izquierdo']:
                grafo.add_edge(nodo['codigo'], nodo['izquierdo']['codigo'])
                construir_grafo(nodo['izquierdo'], x - desplazamiento_x, y - 1, nivel + 1)
                
            if nodo['derecho']:
                grafo.add_edge(nodo['codigo'], nodo['derecho']['codigo'])
                construir_grafo(nodo['derecho'], x + desplazamiento_x, y - 1, nivel + 1)

    construir_grafo(raiz)
    plt.figure(figsize=(8, 5))
    plt.title(titulo, fontsize=14)
    nx.draw(grafo, posiciones, with_labels=True, arrows=False, node_size=1500, node_color="lightblue", font_weight="bold")
    plt.show()

def recolectar_datos():
    """Solicita al usuario la cantidad de nodos y los valida de forma estricta."""
    while True:
        try:
            n = int(input("Ingrese la cantidad de códigos (N): "))
            if n <= 0:
                print("[!] Error: La cantidad debe ser un número entero mayor a 0.")
                continue
            break
        except ValueError:
            print("[!] Error: Entrada no válida. Por favor, ingrese un número entero.")

    codigos = []
    print(f"\nIngrese los {n} códigos (pueden ser separados por espacio o línea por línea):")
    
    # 2. Validar cada código ingresado
    while len(codigos) < n:
        entrada = input(f"  > Faltan {n - len(codigos)} código(s): ").strip().split()
        for item in entrada:
            if len(codigos) < n:
                try:
                    valor = int(item)
                    codigos.append(valor)
                except ValueError:
                    print(f"    [!] Advertencia: '{item}' no es un número válido y será ignorado.")
    
    return codigos

def main():
    print("=" * 45)
    print("      GESTIÓN DE ÁRBOLES BST Y AVL")
    print("=" * 45)

    # 1. Capturar códigos del usuario
    codigos = recolectar_datos()
    print(f"\nLista final de códigos a procesar: {codigos}\n")

    # 2. Inicializar árboles vacíos
    raiz_bst = None
    raiz_avl = None

    # 3. Procesar e insertar en ambos árboles
    for codigo in codigos:
        raiz_bst = insertar_bst(raiz_bst, codigo)
        raiz_avl = insertar_avl(raiz_avl, codigo)

    # 4. Graficar las estructuras resultantes
    print("Mostrando visualizaciones gráficas...")
    graficar_arbol(raiz_bst, "Árbol Binario de Búsqueda (BST)")
    graficar_arbol(raiz_avl, "Árbol AVL (Auto-balanceado)")

    # 5. Ejecutar y mostrar recorridos recursivos para el BST
    pre_bst, in_bst, post_bst = [], [], []
    recorrido_preorden(raiz_bst, pre_bst)
    recorrido_inorden(raiz_bst, in_bst)
    recorrido_postorden(raiz_bst, post_bst)

    print("-" * 45)
    print("RECORRIDOS RECURSIVOS - BST")
    print("-" * 45)
    print(f"  Preorden  : {pre_bst}")
    print(f"  Inorden   : {in_bst}")
    print(f"  Postorden : {post_bst}")

    # 6. Ejecutar y mostrar recorridos recursivos para el AVL
    pre_avl, in_avl, post_avl = [], [], []
    recorrido_preorden(raiz_avl, pre_avl)
    recorrido_inorden(raiz_avl, in_avl)
    recorrido_postorden(raiz_avl, post_avl)

    print("\n" + "-" * 45)
    print("RECORRIDOS RECURSIVOS - AVL")
    print("-" * 45)
    print(f"  Preorden  : {pre_avl}")
    print(f"  Inorden   : {in_avl}")
    print(f"  Postorden : {post_avl}")
    print("=" * 45)

main()
```
