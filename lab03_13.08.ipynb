## 1) Descripcion del problema:
El dueño de una tienda necesita un programa para gestionar el inventario de productos. Los productos se identifican mediante códigos que se ingresan de forma desordenada. El problema consiste en desarrollar un programa que permita registrar una cantidad n de códigos, ordenarlos utilizando el algoritmo Merge Sort, y finalmente, permitir al usuario verificar la existencia de un producto utilizando el algoritmo de Búsqueda Binaria sobre la lista ya ordenada.

## 8) Codigo Funcional:

```python
def ingresar_datos():
    """
    Función que pide al usuario la cantidad de productos y sus codigos.
    """
    n = 0 
    while n < 1:
        n = int(input("Ingrese la cantidad de productos a registrar: "))
        if n < 1:
            print("Error: La cantidad de productos debe ser mayor o igual a 1.")

    codigos = []
    print("\n--- Ingreso de Codigos ---")
    for rep in range(n):
        codigo = -1
        while codigo < 0:
            codigo = int(input(f"Ingrese el codigo del producto {rep + 1}: "))
            if codigo < 0:
                print("Error: El codigo no puede ser un valor negativo.")
        codigos.append(codigo)               
    return codigos
def solicitar_codigo():
    """
    Función que pide al usuario el codigo a verificar.
    """
    codigo = 0 
    while codigo < 1:
        codigo = int(input("Ingrese el codigo a buscar: "))
        if codigo < 1:
            print("Error: el codigo debe ser mayor o igual a 1.")
    return codigo
def merge_sort(codigos):
    """
    Funcion que ordena el arreglo codigos con el algoritmo de merge sort
    """
    if len(codigos) > 1:
        medio = len(codigos) // 2
        izquierda = codigos[:medio]
        derecha = codigos[medio:]
        merge_sort(izquierda)
        merge_sort(derecha)
        i = 0
        j = 0
        k = 0
        while i < len(izquierda) and j < len(derecha):
            if izquierda[i] < derecha[j]:
                codigos[k] = izquierda[i]
                i += 1
            else:
                codigos[k] = derecha[j]
                j += 1
            k += 1
        while i < len(izquierda):
            codigos[k] = izquierda[i]
            i += 1
            k += 1
        while j < len(derecha):
            codigos[k] = derecha[j]
            j += 1
            k += 1
    return codigos
def busqueda_binaria(codigos, codigo):
    inicio = 0
    fin = len(codigos) - 1

    while inicio <= fin:
        medio = (inicio + fin) // 2
        
        if codigos[medio] == codigo:
            return True
        elif codigos[medio] < codigo:
            inicio = medio + 1
        else:
            fin = medio - 1
    return False

def main():
    arreglo_codigos = ingresar_datos()
    print(f"Lista de codigos inicial: {arreglo_codigos}")
    arreglo_codigos_ordenado = merge_sort(arreglo_codigos)
    print(f"Lista de codigos ordenados: {arreglo_codigos_ordenado}")
    codigo_a_buscar = solicitar_codigo()
    encontrado = busqueda_binaria(arreglo_codigos_ordenado, codigo_a_buscar)
    print("\n--- Resultado de la Búsqueda ---")
    if encontrado:
        print(f"El producto con código {codigo_a_buscar} fue encontrado.")
    else:
        print(f"El producto con código {codigo_a_buscar} no se encuentra registrado.")
main()
```
